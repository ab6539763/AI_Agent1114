## 📚 课堂核心技术笔记

### 1. 1. 国内开发环境加速：pip 镜像源与 Git 规范

国内开发安装 Python 第三方库，务必第一时间配置国内镜像源，常用清华源：`pip config set global.index-url https://pypi.tuna.tsinghua.edu.cn/simple`。同时配置 Git 身份信息（`git config --global user.name` 与 `user.email`），每次提交遵循 `feat(day01): 描述` 的格式，方便团队回溯每次迭代。

> **⚠️ 老陈避坑**：常见坑：Windows 安装 Python 时忘了勾选「Add Python to PATH」，导致终端输入 python 提示找不到命令；或者系统自带 Python2 与 Python3 混用，建议统一明确使用 python3 / pip3。

### 2. 2. 变量命名与四大核心数据类型

Python 采用蛇形命名法（snake_case），例如 `employee_id`、`monthly_budget`；常量用全大写如 `DEFAULT_QUOTA = 50000`。四大核心基础类型：`int`（整数，支持任意精度）、`float`（浮点数，注意精度误差）、`str`（Unicode 字符串，支持中文）、`bool`（True/False，本质是 int 的子类）。可用 `type(x)` 查看类型，用 `isinstance(x, int)` 做安全类型判断。

> **⚠️ 老陈避坑**：常见坑：Python 是动态强类型语言，`'工号:' + 1001` 会直接抛出 `TypeError`！必须显式转成 `str(1001)` 或直接用 f-string：`f'工号:{1001}'`。此外 `bool('False')` 结果是 `True`，因为非空字符串转布尔永远为真！

### 3. 3. Git 四步日常提交法

每天在星辰数科写完代码，标准四步：`git checkout -b feat/dayXX-xxx` 切新分支 -> `git status` 检查改动 -> `git add .` 暂存 -> `git commit -m 'feat(dayXX): ...'` 提交 -> `git push -u origin 分支名` 发起合并请求。

> **⚠️ 老陈避坑**：千万不要把 `.venv/` 虚拟环境目录、`.env` 密钥文件或 `__pycache__/` 提交到 Git 仓库！一定要先配置好 `.gitignore`。


## 💻 实操步骤 1：步骤一：编写星辰智脑开发环境自检脚本

- **目标文件**：`versions/day01/src/day01_env_check.py`
- **老陈代码旁白**：先别急着写业务，大厂规范是先跑环境健康检查。咱们写一个小工具，自动检测当前 Python 版本是否满足 3.10+ 要求，并打印清华镜像源配置指令。


````python
"""
STARMIND-001 模块一：开发环境健康自检工具。
检查 Python 运行时版本、操作系统信息，并输出推荐的国内 pip 镜像配置命令。
"""

import platform
import sys
from typing import Dict, Any

# 定义星辰智脑项目要求的最低 Python 主次版本
MIN_PYTHON_MAJOR = 3
MIN_PYTHON_MINOR = 10
TSINGHUA_PYPI_MIRROR = "https://pypi.tuna.tsinghua.edu.cn/simple"


def check_environment() -> Dict[str, Any]:
    """收集当前开发环境信息并判断是否满足项目基线要求。"""
    # 获取当前 Python 解释器的主版本、次版本、微版本号
    version_info = sys.version_info
    version_str = f"{version_info.major}.{version_info.minor}.{version_info.micro}"

    # 判断是否达到 Python 3.10 及以上版本
    is_compatible = (version_info.major, version_info.minor) >= (
        MIN_PYTHON_MAJOR,
        MIN_PYTHON_MINOR,
    )

    # 组装结构化的诊断结果字典
    report: Dict[str, Any] = {
        "python_version": version_str,
        "os_name": platform.system(),
        "os_release": platform.release(),
        "is_compatible": is_compatible,
        "recommended_mirror": TSINGHUA_PYPI_MIRROR,
        "pip_config_cmd": f"pip config set global.index-url {TSINGHUA_PYPI_MIRROR}",
    }
    return report


def format_env_report(report: Dict[str, Any]) -> str:
    """将环境诊断结果格式化为易读的终端输出文本。"""
    status_badge = "[PASS] 满足星辰智脑运行要求" if report["is_compatible"] else "[WARN] 请升级至 Python 3.10+"
    lines = [
        "=== 星辰智脑 StarMind 开发环境自检报告 ===",
        f"Python 版本 : {report['python_version']}  {status_badge}",
        f"操作系统    : {report['os_name']} ({report['os_release']})",
        f"推荐镜像源  : {report['recommended_mirror']}",
        f"一键配置命令: {report['pip_config_cmd']}",
        "============================================",
    ]
    return "\n".join(lines)


if __name__ == "__main__":
    # 执行环境自检并打印报告
    env_report = check_environment()
    print(format_env_report(env_report))
````

### 🔍 步骤 1 代码深度解析

1. 使用标准库 `sys.version_info` 而不是用字符串去比较版本号（因为字符串比较 `'3.9' > '3.11'` 会出错，元组比较 `(3, 11) >= (3, 10)` 才准确）。
2. 把「数据采集 `check_environment()`」与「文本展示 `format_env_report()`」分开写，这样单元测试可以直接断言 `report['is_compatible'] is True`。

## 💻 实操步骤 2：步骤二：实现星辰员工数字名片生成器（含手机号脱敏与配额计算）

- **目标文件**：`versions/day01/src/day01_profile_card.py`
- **老陈代码旁白**：环境没问题了，咱们直接搞定林产品的 STARMIND-001 需求。注意手机号必须做中间四位脱敏，AI Token 初始配额根据工龄动态计算。


````python
"""
STARMIND-001 模块二：星辰数科新员工数字名片生成器。
演示变量定义、四大基础类型（str/int/float/bool）、类型转换与安全脱敏输出。
"""

# 常量定义：使用全大写蛇形命名法
COMPANY_NAME = "杭州星辰数科"
PLATFORM_NAME = "星辰智脑 StarMind"
BASE_DAILY_TOKEN_QUOTA = 50000  # 每位新员工基础每日 Token 配额


def mask_phone(phone: str) -> str:
    """对 11 位中国大陆手机号进行中间四位脱敏处理。"""
    # 去除首尾可能存在的空格
    cleaned = str(phone).strip()
    # 校验长度是否为 11 位且全为数字
    if len(cleaned) == 11 and cleaned.isdigit():
        # 保留前 3 位和后 4 位，中间替换为 ****
        return f"{cleaned[:3]}****{cleaned[-4:]}"
    # 非标准手机号则返回统一占位，避免明文泄露
    return "格式异常(已隐藏)"


def calculate_token_quota(years_of_exp: float, is_ai_team: bool) -> int:
    """根据工作年限与是否属于 AI 研发团队计算每日 Token 配额。"""
    # 每增加 1 年经验额外增加 10000 Token 配额，先按浮点计算再转为整数
    exp_bonus = int(max(0.0, years_of_exp) * 10000)
    # AI 核心研发团队享有 1.5 倍配额加成
    multiplier = 1.5 if is_ai_team else 1.0
    total_quota = int((BASE_DAILY_TOKEN_QUOTA + exp_bonus) * multiplier)
    return total_quota


def build_profile_card(
    name: str,
    employee_id: int,
    department: str,
    phone: str,
    years_of_exp: float,
    is_ai_team: bool = True,
) -> str:
    """构建标准格式的员工数字名片字符串。"""
    # 格式化 4 位补零工号，例如 7 -> "XM-0007"
    formatted_id = f"XM-{int(employee_id):04d}"
    # 执行手机号脱敏
    safe_phone = mask_phone(phone)
    # 计算每日 Token 配额
    quota = calculate_token_quota(years_of_exp, is_ai_team)
    # 将布尔值转换为中文权限描述
    ai_status = "已开通 (核心研发权限)" if is_ai_team else "已开通 (标准办公权限)"

    # 拼装多行名片文本，利用 f-string 格式化千分位逗号与保留 1 位小数
    card = (
        f"+------------------------------------------+\n"
        f"|  {COMPANY_NAME} · {PLATFORM_NAME} 数字名片  |\n"
        f"+------------------------------------------+\n"
        f"  姓名     : {name}\n"
        f"  工号     : {formatted_id}\n"
        f"  部门     : {department}\n"
        f"  联系电话 : {safe_phone}\n"
        f"  工作年限 : {years_of_exp:.1f} 年\n"
        f"  AI 权限  : {ai_status}\n"
        f"  每日配额 : {quota:,} Tokens\n"
        f"+------------------------------------------+"
    )
    return card


if __name__ == "__main__":
    # 模拟新员工「陈小新」入职生成数字名片（不阻塞 input，方便直接运行验证）
    demo_card = build_profile_card(
        name="陈小新",
        employee_id=42,
        department="AI 应用研发部",
        phone="13857108899",
        years_of_exp=0.5,
        is_ai_team=True,
    )
    print(demo_card)
````

### 🔍 步骤 2 代码深度解析

1. `f'XM-{int(employee_id):04d}'` 中的 `:04d` 表示不足 4 位自动左侧补 0，企业工号、订单号极常用。
2. `f'{quota:,}'` 自动加上千分位逗号（如 `82,500`），打印报表和 Token 账单时非常清爽。
3. `max(0.0, years_of_exp)` 属于防御性编程：万一上游传入负数工龄，不会把配额扣成负数。

### 作业 1（基础巩固）：基础巩固：API 费用人民币估算器

**题目要求：**
假设星辰智脑调用 DeepSeek 模型输入单价为 0.002 元/千 Tokens，输出单价为 0.008 元/千 Tokens。编写函数 `estimate_cost(input_tokens: int, output_tokens: int) -> float`，返回本次对话的总费用（保留 6 位小数），并打印格式化账单。

**思路提示：** 注意单位换算是除以 1000.0，最后使用 round(cost, 6) 消除浮点尾部噪音。


````python
"""Day 1 课后题 1：大模型调用费用估算器。"""

INPUT_PRICE_PER_1K = 0.002   # 输入单价：元 / 1000 Tokens
OUTPUT_PRICE_PER_1K = 0.008  # 输出单价：元 / 1000 Tokens


def estimate_cost(input_tokens: int, output_tokens: int) -> float:
    """计算单次大模型对话的人民币费用（元）。"""
    # 防止传入负数 Token
    safe_in = max(0, int(input_tokens))
    safe_out = max(0, int(output_tokens))
    # 按千 Token 单价分别计算并相加
    total = (safe_in / 1000.0) * INPUT_PRICE_PER_1K + (safe_out / 1000.0) * OUTPUT_PRICE_PER_1K
    return round(total, 6)


if __name__ == "__main__":
    cost = estimate_cost(1500, 800)
    print(f"本次调用输入 1500 Tokens，输出 800 Tokens，预计费用：¥{cost:.6f}")
````

**答案解析与评分标准：**

重点考查 int/float 混合运算与 `round()` 精度控制。评分要点：1）常量抽取清晰；2）边界负数保护；3）格式化输出保留 6 位小数。


### 作业 2（企业实战）：企业实战：多字段员工隐私脱敏工具

**题目要求：**
林产品反馈除了手机号，员工邮箱（如 `chenxiaoxin@starmind.cn`）和身份证号也需要脱敏。编写 `mask_email(email: str) -> str`（保留首字符与 `@` 后域名，中间用 `***` 替代），并在主流程中输出完整的脱敏摘要。

**思路提示：** 用 `email.split('@')` 拆出用户名和域名，先判断 `'@' in email` 防止报错。


````python
"""Day 1 课后题 2：企业员工邮箱与身份信息脱敏工具。"""


def mask_email(email: str) -> str:
    """对企业邮箱执行脱敏：保留前缀首字母与完整域名。"""
    cleaned = email.strip()
    if "@" not in cleaned:
        return "无效邮箱"
    local_part, domain = cleaned.split("@", 1)
    if not local_part or not domain:
        return "无效邮箱"
    # 保留用户名第一个字符，其余用 *** 掩码
    return f"{local_part[0]}***@{domain}"


if __name__ == "__main__":
    for sample in ["chenxiaoxin@starmind.cn", "a@corp.com", "invalid-mail"]:
        print(f"{sample:25s} -> {mask_email(sample)}")
````

**答案解析与评分标准：**

考查字符串切分 `split('@', 1)` 与异常输入防御。注意限制 `maxsplit=1`，防止异常字符串包含多个 `@` 导致解包报错 `ValueError`。


### 作业 3（架构进阶）：架构进阶：带环境变量检测的启动预检器

**题目要求：**
在企业服务启动前，往往需要检查关键配置（如 `STARMIND_ENV`、`PYTHONUTF8`）是否就绪。编写 `preflight_check(env_map: dict) -> tuple[bool, str]`，在不依赖真实操作系统的情况下支持注入字典做可测试的启动预检。

**思路提示：** 通过参数传入 `env_map` 而不是硬编码 `os.environ`，这就是架构设计中的「依赖注入」雏形。


````python
"""Day 1 课后题 3：可测试的服务启动预检器（依赖注入雏形）。"""

from typing import Dict, Tuple


def preflight_check(env_map: Dict[str, str]) -> Tuple[bool, str]:
    """检查运行环境配置是否满足星辰智脑启动条件。"""
    # 检查运行环境标识是否合法
    app_env = env_map.get("STARMIND_ENV", "dev").lower()
    valid_envs = ("dev", "staging", "prod")
    if app_env not in valid_envs:
        return False, f"非法环境标识: {app_env}，仅允许 {valid_envs}"

    # 生产环境强制要求配置 API 密钥
    if app_env == "prod" and not env_map.get("STARMIND_API_KEY"):
        return False, "生产环境缺少 STARMIND_API_KEY，拒绝启动！"

    return True, f"预检通过，当前模式: {app_env}"


if __name__ == "__main__":
    print(preflight_check({"STARMIND_ENV": "dev"}))
    print(preflight_check({"STARMIND_ENV": "prod"}))
    print(preflight_check({"STARMIND_ENV": "prod", "STARMIND_API_KEY": "sk-demo"}))
````

**答案解析与评分标准：**

本题提前引入「依赖注入」思想：把外部环境变量作为函数参数传入，测试时不用修改真实系统环境变量就能覆盖 dev/prod/缺密钥等全部场景。
